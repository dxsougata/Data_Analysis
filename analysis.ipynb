```python
import matplotlib.pyplot as plt
```

```python
df["Target"].value_counts().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Distribution of Student Outcomes")
plt.xlabel("Student Outcome")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Distribution of age
plt.figure(figsize=(8, 5))

plt.hist(df["Age at enrollment"], bins=20)

plt.title("Distribution of Age at Enrollment")
plt.xlabel("Age at Enrollment")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
#Distribution of admission grade 

plt.figure(figsize=(8, 5))

plt.hist(df["Admission grade"], bins=20)

plt.title("Distribution of Admission Grade")
plt.xlabel("Admission Grade")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
#Box plot for detecting outlier for age 

plt.figure(figsize=(8, 4))

plt.boxplot(df["Age at enrollment"])

plt.title("Box Plot of Age at Enrollment")
plt.ylabel("Age")

plt.tight_layout()
plt.show()
```

```python
#box plot for detecting outlier for admission grade

plt.figure(figsize=(8, 4))

plt.boxplot(df["Admission grade"])

plt.title("Box Plot of Admission Grade")
plt.ylabel("Admission Grade")

plt.tight_layout()
plt.show()
```

```python
#analyze academic performance

academic_cols = [
    "Curricular units 1st sem (grade)",
    "Curricular units 2nd sem (grade)"
]

print(df[academic_cols].describe())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Curricular units 1st sem (grade)"], bins=20)

plt.title("Distribution of First Semester Grades")
plt.xlabel("First Semester Grade")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Curricular units 2nd sem (grade)"], bins=20)

plt.title("Distribution of Second Semester Grades")
plt.xlabel("Second Semester Grade")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
numeric_df = df.select_dtypes(include=np.number)

print("Numerical columns:")
print(numeric_df.columns.tolist())
```

```python
# correlation analysis


corr = numeric_df.corr()

plt.figure(figsize=(14, 10))

plt.imshow(corr, aspect="auto")
plt.colorbar()

plt.title("Correlation Matrix of Numerical Variables")

plt.tight_layout()
plt.show()
```

```python
# compare academic performance with student outcomes

print(
    df.groupby("Target")["Curricular units 1st sem (grade)"]
    .mean()
)
```

```python
print(
    df.groupby("Target")["Curricular units 2nd sem (grade)"]
    .mean()
)
```

```python
df.boxplot(
    column="Curricular units 1st sem (grade)",
    by="Target",
    figsize=(8, 5)
)

plt.title("First Semester Grades by Student Outcome")
plt.suptitle("")
plt.xlabel("Student Outcome")
plt.ylabel("First Semester Grade")

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(14, 10))

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.show()
```
