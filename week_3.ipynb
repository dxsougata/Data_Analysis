```python

# ================================================================
# WEEK 3 TASK: DATA VISUALIZATION AND REPORTING
# Dataset: Student Dropout and Academic Success
# Libraries: Pandas, NumPy, Matplotlib, Seaborn
# Run this entire block in ONE Jupyter Notebook cell.
# ================================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

warnings.filterwarnings("ignore")

# -------------------- 1. VISUALIZATION SETTINGS --------------------

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 11
plt.rcParams["figure.dpi"] = 110

OUTPUT_DIR = "week3_visualizations"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 70)
print("WEEK 3: DATA VISUALIZATION AND REPORTING")
print("=" * 70)

# -------------------- 2. LOAD THE DATASET --------------------

possible_files = [
    "data.csv",
    "student_data_cleaned.csv",
    "student_data.csv"
]

file_path = next(
    (f for f in possible_files if os.path.exists(f)),
    None
)

if file_path is None:
    raise FileNotFoundError(
        "Dataset not found. Place data.csv or student_data.csv "
        "in the same folder as your Jupyter Notebook."
    )

# sep=None automatically detects common delimiters such as commas
# and semicolons.
df = pd.read_csv(file_path, sep=None, engine="python", encoding="utf-8-sig")

# Standardize column names by removing unwanted spaces and tabs.
df.columns = (
    df.columns.astype(str)
    .str.replace("\ufeff", "", regex=False)
    .str.strip()
)

print(f"\nDataset loaded successfully: {file_path}")
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
print("\nFirst five records:")
display(df.head())

print("\nColumn data types:")
display(df.dtypes.rename("Data Type").to_frame())

print("\nMissing values in each column:")
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

if missing.empty:
    print("No missing values were detected.")
else:
    display(missing.to_frame("Missing Values"))

# -------------------- 3. IDENTIFY IMPORTANT COLUMNS --------------------

# Use the Target column when available.
target_col = next(
    (c for c in df.columns if c.lower() == "target"),
    None
)

if target_col is None:
    target_col = next(
        (c for c in df.columns
         if any(word in c.lower()
                for word in ["outcome", "status", "result", "class"])),
        None
    )

if target_col is None:
    raise ValueError(
        "No target/outcome column was detected. "
        "Please check the dataset's column names."
    )

# Convert numeric-looking columns where appropriate.
for col in df.columns:
    if col != target_col and df[col].dtype == "object":
        converted = pd.to_numeric(df[col], errors="coerce")
        if converted.notna().sum() >= 0.9 * df[col].notna().sum():
            df[col] = converted

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
numeric_cols = [c for c in numeric_cols if c != target_col]

# Target is categorical for this analysis.
df[target_col] = df[target_col].astype("string").str.strip()
df.loc[df[target_col] == "", target_col] = pd.NA

# Work with valid target records for grouped plots.
plot_df = df.dropna(subset=[target_col]).copy()

print(f"\nOutcome column: {target_col}")
print(f"Numeric predictor columns: {len(numeric_cols)}")
print(f"Records available for outcome analysis: {len(plot_df)}")

print("\nOutcome distribution:")
target_counts = plot_df[target_col].value_counts()
target_summary = pd.DataFrame({
    "Count": target_counts,
    "Percentage": (target_counts / target_counts.sum() * 100).round(2)
})
display(target_summary)

# Helper: save and display each figure separately.
def show_plot(filename):
    path = os.path.join(OUTPUT_DIR, filename)
    plt.tight_layout()
    plt.savefig(path, dpi=200, bbox_inches="tight")
    plt.show()
    print(f"Saved visualization: {path}\n")


# -------------------- 4. BAR CHART --------------------
# Question: How many students belong to each outcome category?

print("\n" + "=" * 70)
print("VISUALIZATION 1: BAR CHART — STUDENT OUTCOMES")
print("=" * 70)

plt.figure(figsize=(9, 5))
ax = sns.countplot(
    data=plot_df,
    x=target_col,
    order=target_counts.index,
    hue=target_col,
    legend=False
)

ax.set_title("Distribution of Student Outcomes")
ax.set_xlabel("Student Outcome")
ax.set_ylabel("Number of Students")
ax.tick_params(axis="x", rotation=15)

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=3)

show_plot("01_student_outcomes_bar.png")

majority_outcome = target_counts.index[0]
majority_percentage = target_counts.iloc[0] / target_counts.sum() * 100

print("INTERPRETATION:")
print(f"- The largest outcome category is '{majority_outcome}'.")
print(f"- It represents {majority_percentage:.2f}% of records.")
print("- Differences in bar heights show whether the outcome classes "
      "are balanced.")
print("- If one class dominates, predictive models may need class-wise "
      "evaluation rather than accuracy alone.")


# -------------------- 5. LINE CHART --------------------
# Question: How do academic performance indicators vary by outcome?
# Prefer meaningful academic indicators when available.

def find_column(possible_names):
    normalized = {c.lower().strip(): c for c in df.columns}

    for name in possible_names:
        if name.lower() in normalized:
            return normalized[name.lower()]

    for c in df.columns:
        cl = c.lower()
        if any(name.lower() in cl for name in possible_names):
            return c

    return None


preferred_line_cols = [
    find_column(["Previous qualification (grade)"]),
    find_column(["Admission grade"]),
    find_column(["Curricular units 1st sem (grade)"]),
    find_column(["Curricular units 2nd sem (grade)"])
]

line_cols = []
for c in preferred_line_cols:
    if c and c in numeric_cols and c not in line_cols:
        line_cols.append(c)

# Fallback if academic-grade columns are unavailable.
if len(line_cols) < 2:
    fallback = [
        c for c in numeric_cols
        if plot_df[c].nunique(dropna=True) > 1
    ]
    line_cols = list(dict.fromkeys(line_cols + fallback))[:5]

line_cols = [
    c for c in line_cols
    if plot_df[c].notna().sum() > 0
]

print("\n" + "=" * 70)
print("VISUALIZATION 2: LINE CHART — ACADEMIC INDICATORS BY OUTCOME")
print("=" * 70)

if len(line_cols) >= 2:
    line_data = (
        plot_df.groupby(target_col, observed=True)[line_cols]
        .mean()
        .reindex(target_counts.index)
    )

    plt.figure(figsize=(12, 6))

    for col in line_cols:
        plt.plot(
            range(len(line_data)),
            line_data[col],
            marker="o",
            linewidth=2,
            label=col
        )

    plt.xticks(
        range(len(line_data)),
        line_data.index.astype(str),
        rotation=15
    )
    plt.title("Mean Academic Indicators Across Student Outcomes")
    plt.xlabel("Student Outcome")
    plt.ylabel("Mean Value (Original Units)")
    plt.legend(
        title="Academic Indicator",
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )
    show_plot("02_academic_indicators_line.png")

    print("INTERPRETATION:")
    print("- Each line represents the average value of one selected "
          "numeric indicator across outcome categories.")
    print("- Compare the direction and separation of the lines to "
          "identify indicators that differ between outcomes.")
    print("- These are group averages, not proof that a particular "
          "indicator causes dropout or graduation.")
    display(line_data.round(2))
else:
    print("Not enough usable numeric columns for a meaningful line chart.")


# -------------------- 6. SCATTER PLOT --------------------
# Question: Is academic progress in one semester associated with
# progress in another semester?

x_preferred = find_column([
    "Curricular units 1st sem (approved)"
])
y_preferred = find_column([
    "Curricular units 2nd sem (approved)"
])

if (
    x_preferred in numeric_cols
    and y_preferred in numeric_cols
    and x_preferred != y_preferred
):
    x_col, y_col = x_preferred, y_preferred
else:
    scatter_candidates = [
        c for c in numeric_cols
        if plot_df[c].nunique(dropna=True) > 1
    ]
    x_col, y_col = (
        scatter_candidates[:2]
        if len(scatter_candidates) >= 2
        else (None, None)
    )

print("\n" + "=" * 70)
print("VISUALIZATION 3: SCATTER PLOT — RELATIONSHIP BETWEEN VARIABLES")
print("=" * 70)

if x_col and y_col:
    scatter_df = plot_df[[x_col, y_col, target_col]].dropna()

    plt.figure(figsize=(10, 6))
    sns.scatterplot(
        data=scatter_df,
        x=x_col,
        y=y_col,
        hue=target_col,
        alpha=0.65,
        s=55
    )
    plt.title(f"{y_col} vs {x_col}, Grouped by Student Outcome")
    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.legend(title="Student Outcome", bbox_to_anchor=(1.02, 1),
               loc="upper left")
    show_plot("03_relationship_scatter.png")

    if len(scatter_df) > 2:
        corr_value = scatter_df[x_col].corr(scatter_df[y_col])
        print("INTERPRETATION:")
        print(f"- Variables compared: {x_col} and {y_col}.")
        print(f"- Pearson correlation: {corr_value:.3f}")
        print("- A positive correlation indicates that larger values "
              "of one variable tend to accompany larger values of the other.")
        print("- The color groups help reveal whether the outcomes occupy "
              "different regions of the scatter plot.")
        print("- Correlation does not establish causation.")
else:
    print("Not enough usable numeric columns for a scatter plot.")


# -------------------- 7. CORRELATION HEATMAP --------------------
# Question: Which numeric variables move together?

print("\n" + "=" * 70)
print("VISUALIZATION 4: HEATMAP — NUMERIC CORRELATIONS")
print("=" * 70)

usable_numeric = [
    c for c in numeric_cols
    if plot_df[c].notna().sum() > 1
    and plot_df[c].nunique(dropna=True) > 1
]

if len(usable_numeric) >= 2:
    # Limit the heatmap to the 15 most variable numeric features
    # so that labels remain readable.
    variance_rank = (
        plot_df[usable_numeric].var(numeric_only=True)
        .sort_values(ascending=False)
    )
    heat_cols = variance_rank.head(15).index.tolist()

    corr_matrix = plot_df[heat_cols].corr()

    plt.figure(figsize=(13, 10))
    sns.heatmap(
        corr_matrix,
        annot=len(heat_cols) <= 10,
        fmt=".2f",
        cmap="coolwarm",
        center=0,
        vmin=-1,
        vmax=1,
        linewidths=0.4,
        square=True
    )
    plt.title("Correlation Heatmap of Numeric Features")
    show_plot("04_correlation_heatmap.png")

    print("INTERPRETATION:")
    print("- Correlation values range from -1 to +1.")
    print("- Values near +1 indicate a strong positive linear association.")
    print("- Values near -1 indicate a strong negative linear association.")
    print("- Values near 0 indicate a weak linear relationship.")
    print("- Strongly correlated predictors may contain overlapping "
          "information, but domain knowledge is needed to interpret them.")

    upper_triangle = corr_matrix.where(
        np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
    )
    strongest = (
        upper_triangle.stack()
        .sort_values(key=lambda s: s.abs(), ascending=False)
    )

    print("\nStrongest feature correlations:")
    display(strongest.head(10).rename("Correlation").to_frame())
else:
    print("Not enough variable numeric columns for a correlation heatmap.")


# -------------------- 8. BOXPLOT: OUTLIER INVESTIGATION --------------------
# Question: Do numeric variables contain unusually extreme values?

print("\n" + "=" * 70)
print("VISUALIZATION 5: BOXPLOT — POTENTIAL OUTLIERS")
print("=" * 70)

if usable_numeric:
    box_cols = (
        plot_df[usable_numeric].var(numeric_only=True)
        .sort_values(ascending=False)
        .head(min(6, len(usable_numeric)))
        .index.tolist()
    )

    box_data = plot_df[box_cols].melt(
        var_name="Feature",
        value_name="Value"
    ).dropna()

    plt.figure(figsize=(13, 6))
    sns.boxplot(data=box_data, x="Feature", y="Value")
    plt.title("Distribution and Potential Outliers in Numeric Features")
    plt.xlabel("Numeric Feature")
    plt.ylabel("Value (Original Units)")
    plt.xticks(rotation=35, ha="right")
    show_plot("05_numeric_boxplots.png")

    print("INTERPRETATION:")
    print("- The box shows the interquartile range, while the center "
          "line indicates the median.")
    print("- Points beyond the whiskers may be potential outliers.")
    print("- An extreme observation is not automatically an error; "
          "it should be investigated before removal.")
    print("- The selected features have different units, so compare "
          "their distributions individually.")
else:
    print("No usable numeric features for boxplots.")


# -------------------- 9. HISTOGRAM: FEATURE DISTRIBUTIONS --------------------
# Question: Are numeric variables skewed or concentrated in certain ranges?

print("\n" + "=" * 70)
print("VISUALIZATION 6: HISTOGRAMS — FEATURE DISTRIBUTIONS")
print("=" * 70)

if usable_numeric:
    hist_cols = (
        plot_df[usable_numeric].var(numeric_only=True)
        .sort_values(ascending=False)
        .head(min(4, len(usable_numeric)))
        .index.tolist()
    )

    fig, axes = plt.subplots(
        nrows=len(hist_cols),
        ncols=1,
        figsize=(10, 4 * len(hist_cols))
    )
    axes = np.atleast_1d(axes)

    for ax, col in zip(axes, hist_cols):
        sns.histplot(
            data=plot_df,
            x=col,
            bins=25,
            kde=True,
            ax=ax
        )
        ax.set_title(f"Distribution of {col}")
        ax.set_xlabel(col)
        ax.set_ylabel("Frequency")

    show_plot("06_feature_histograms.png")

    print("INTERPRETATION:")
    print("- Histograms show the frequency of values across intervals.")
    print("- A symmetric shape suggests a more balanced distribution.")
    print("- A long tail can indicate skewness.")
    print("- Multiple peaks may suggest subgroups or distinct patterns.")
    print("- These patterns help decide whether transformations are "
          "worth testing before statistical modeling.")


# -------------------- 10. OUTCOME-WISE ACADEMIC COMPARISON --------------------
# Question: How do semester performance indicators differ by outcome?

print("\n" + "=" * 70)
print("VISUALIZATION 7: GROUP COMPARISON — SEMESTER PERFORMANCE")
print("=" * 70)

sem1 = find_column(["Curricular units 1st sem (grade)"])
sem2 = find_column(["Curricular units 2nd sem (grade)"])

grade_cols = [
    c for c in [sem1, sem2]
    if c and c in numeric_cols
]

if grade_cols:
    grade_data = plot_df[[target_col] + grade_cols].melt(
        id_vars=target_col,
        var_name="Semester Indicator",
        value_name="Grade"
    ).dropna()

    plt.figure(figsize=(11, 6))
    sns.boxplot(
        data=grade_data,
        x=target_col,
        y="Grade",
        hue="Semester Indicator"
    )
    plt.title("Semester Grades Across Student Outcomes")
    plt.xlabel("Student Outcome")
    plt.ylabel("Grade")
    plt.legend(title="Semester Indicator")
    plt.xticks(rotation=15)
    show_plot("07_semester_grade_comparison.png")

    print("INTERPRETATION:")
    print("- Compare the median grades and spread for each outcome.")
    print("- Higher medians can indicate stronger academic performance "
          "within a group.")
    print("- Overlapping boxes indicate that the distributions share "
          "some ranges.")
    print("- Grade differences may be associated with student outcomes, "
          "but they do not alone explain every student's result.")
else:
    print("Semester grade columns were not found; this plot was skipped.")


# -------------------- 11. OPTIONAL: CATEGORICAL COMPARISON --------------------
# Compare tuition status with student outcome, when available.

tuition_col = find_column(["Tuition fees up to date"])

print("\n" + "=" * 70)
print("VISUALIZATION 8: TUITION STATUS AND STUDENT OUTCOME")
print("=" * 70)

if tuition_col:
    tuition_data = plot_df[[tuition_col, target_col]].dropna().copy()
    tuition_data[tuition_col] = tuition_data[tuition_col].astype(str)

    plt.figure(figsize=(10, 6))
    sns.countplot(
        data=tuition_data,
        x=tuition_col,
        hue=target_col
    )
    plt.title("Student Outcomes by Tuition Fee Status")
    plt.xlabel("Tuition Fees Up to Date (Encoded Value)")
    plt.ylabel("Number of Students")
    plt.legend(title="Student Outcome", bbox_to_anchor=(1.02, 1),
               loc="upper left")
    show_plot("08_tuition_status_outcomes.png")

    print("INTERPRETATION:")
    print("- This chart compares the counts of each student outcome "
          "across tuition-status categories.")
    print("- Differences can identify patterns worth investigating.")
    print("- The values are encoded categories; interpret their meanings "
          "using the dataset's data dictionary.")
else:
    print("Tuition-status column not found; this plot was skipped.")


# -------------------- 12. SUMMARY TABLES --------------------

print("\n" + "=" * 70)
print("ADDITIONAL ANALYTICAL SUMMARY")
print("=" * 70)

print("\nNumeric descriptive statistics:")
if numeric_cols:
    display(plot_df[numeric_cols].describe().T.round(2))

print("\nMean numeric values by student outcome:")
if numeric_cols:
    outcome_means = plot_df.groupby(
        target_col, observed=True
    )[numeric_cols].mean()
    display(outcome_means.round(2))

print("\nDataset quality summary:")
quality_summary = pd.DataFrame({
    "Metric": [
        "Total records",
        "Total columns",
        "Numeric columns",
        "Categorical columns",
        "Missing cells",
        "Duplicate rows",
        "Outcome categories"
    ],
    "Value": [
        len(df),
        df.shape[1],
        len(df.select_dtypes(include=np.number).columns),
        len(df.select_dtypes(exclude=np.number).columns),
        int(df.isna().sum().sum()),
        int(df.duplicated().sum()),
        int(plot_df[target_col].nunique())
    ]
})
display(quality_summary)

# -------------------- 13. SAVE A SUMMARY REPORT --------------------

summary_path = os.path.join(OUTPUT_DIR, "visualization_summary.txt")

with open(summary_path, "w", encoding="utf-8") as report:
    report.write("WEEK 3: DATA VISUALIZATION AND REPORTING\n")
    report.write("=" * 55 + "\n\n")
    report.write(f"Dataset: {file_path}\n")
    report.write(f"Rows: {df.shape[0]}\n")
    report.write(f"Columns: {df.shape[1]}\n")
    report.write(f"Outcome column: {target_col}\n")
    report.write(f"Missing cells: {df.isna().sum().sum()}\n")
    report.write(f"Duplicate rows: {df.duplicated().sum()}\n\n")

    report.write("Outcome distribution:\n")
    report.write(target_summary.to_string())
    report.write("\n\nNumeric descriptive statistics:\n")

    if numeric_cols:
        report.write(plot_df[numeric_cols].describe().T.to_string())

    report.write("\n\nVisualization files:\n")
    for filename in sorted(os.listdir(OUTPUT_DIR)):
        if filename.endswith(".png"):
            report.write(f"- {filename}\n")

print("\n" + "=" * 70)
print("VISUALIZATION PROCESS COMPLETED")
print("=" * 70)
print(f"All generated charts are saved in: {OUTPUT_DIR}/")
print(f"Summary report saved to: {summary_path}")
print("\nUse the saved chart images and the printed interpretations "
      "when preparing your Week 3 DOC report.")
```
